Prior to analysing the data set, a class "patient" must be created. The following section of code is written in a separate file "patient_Lynch". This is done by first creating an empty list called patients where the items in this class can later be found. Next, a constructor for class patient was created and defined for donorID, sex, whether or not they have dementia, apoe genotype, age of death, and years of education. The representor for the patient class defines what is returned when the class patient is called. Next, the code filters out patients from the list all_patients to return only the patients that meet the listed criteria. 

```python
import csv
class Patient:
    all_patients = []
    #constructor for patient, containing donorID, sex, whether or not they have dementia, apoe genotype, age of death, and years of education
    def __init__(self, donorID:str, sex: str, diagnosis: str, apoe: str, deathage: int, educationyears: int, onsetage: int, brainweight: int, educationlevel: str, mmse: int):
        self.donorID = donorID
        self.sex = sex
        self.diagnosis = diagnosis
        self.apoe = apoe
        self.deathage = deathage
        self.educationyears = educationyears
        self.onsetage = onsetage
        self.brainweight = brainweight
        self.educationlevel = educationlevel
        self.mmse = mmse

        Patient.all_patients.append(self)
    #representor for patient
    def __repr__(self):
        return f"{self.donorID}: ({self.sex} | {self.diagnosis} | {self.apoe} | {self.onsetage} | {self.educationyears})"

    def getAge(self):
        return self.deathage
    #instantiates all patients from csv data
    @classmethod
    def instantiate_from_csv(cls, filename:str):
        with open(filename, encoding="utf8") as f:
            reader = csv.DictReader(f)
            rows_of_patients = list(reader)
            for row in rows_of_patients:
                Patient(donorID = row['Donor ID'], sex = row['Sex'], diagnosis = row['Cognitive Status'], apoe = row['APOE Genotype'], deathage = int(row['Age at Death']), educationyears = int(row['Years of education']), onsetage = int(row['Age of onset cognitive symptoms'].strip()) if row['Age of onset cognitive symptoms'].strip() else None, brainweight = int(row['Fresh Brain Weight']) if row['Fresh Brain Weight'] != "Unavailable" else None, educationlevel = row['Highest level of education'], mmse = int(row['Last MMSE Score'].strip()) if row['Last MMSE Score'].strip() else None)
    #filters list of patients by sex, diagnosis, and/or apoe genotype
    @classmethod
    def filter(cls, list, sex:str = "any", diagnosis:str = "any", apoe:str = "any", onsetage = "any", educationlevel = "any"):
        all_patients = list
        remove_list = []
        attr_list = (sex, diagnosis, apoe, onsetage, educationlevel)
        attr_name = ("sex", "diagnosis", "apoe", "onsetage", "educationlevel")
        for attr in range(len(attr_list)):
            if attr_list[attr] != "any":
                for patient in all_patients:
                    if getattr(patient,attr_name[attr]) !=attr_list[attr]:
                        remove_list.append(patient)
                all_patients = [patient for patient in all_patients if patient not in remove_list]
                remove_list.clear()
        return all_patients
```

The remaining code is written in the main python file. First, the functions needed for the data analysis as well as the data itself must be imported. The code from patient_Lynch must be accessed as well, which is done in the first line of code.

```python
#import the necessary functions and data
from patient_Lynch import *
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import statistics
import pandas as pd
import sklearn
from sklearn.linear_model import LinearRegression

#location of data file
Patient.instantiate_from_csv(r"C:\Users\reill\OneDrive - University of Virginia\BME2315\Module 1\BME2315_Module-1\BME2315_Module-1\Metadata and Protein Data for Module 1.csv")
```

For the first data comparison, a scatterplot with patient age of death and education in years was made. An empty list for each variable was created, and each patient in the dataset was iterated over, with their age of death and years of education added to the corresponding list. Next, the lists containing patient years of education and age of death were assigned a variable (X and y, respectively) and the data was reformatted to fit within a square for easier statistical analysis. 

```python
#creates empty list and iterates over each patient in dataset
patient_deathage = []
patient_educationyears = []
for patient in Patient.all_patients:
    patient_deathage.append(patient.deathage)
for patient in Patient.all_patients:
    patient_educationyears.append(patient.educationyears)

X = np.array(patient_educationyears).reshape(-1,1)
y = np.array(patient_deathage)
```

Once the axes have been defined, the next steps are to create the scatterplot comparing the two variables and conduct the statistical analysis on the plot. To analyze the data, a linear regression function was called, and a line of best fit for X,y of the model was created. The equation of the line of best fit was also annotated on the plot and the labels for the x and y axes and title were created. 

```python
#models a linear regression on the data
model = LinearRegression()
model.fit(X,y)

slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(X, y)

# Annotate equation and plot line
equation = f"y = {slope:.2f}x + {intercept:.2f}\nR² = {r2:.2f}"
plt.text(X.max(), y.max(), equation, color="red", fontsize=12, verticalalignment='top')

plt.scatter (X, y, color='blue')
plt.plot(X, model.predict(X), color='red', linewidth=2)
plt.xlabel('Years of Education')
plt.ylabel('Age of Death')
plt.title('Scatter Plot of Years of Education vs. Age of Death')
plt.show()
```

The next form of anaylsis conducted on the data is in the form of a bar graph. The bar graph compares the years of education in patients with and without dementia. Similarly to the first analysis, two empty lists were created. This time, the lists indicated whether or not the patient had dementia. The code then iterated over each patient, appending them to the appropriate list based on their dementia status. After the patients are sorted, statistic analysis of the data is conducted. This includes the mean number of patients with and without dementia as well as the standard deviation of the same values. 

```python
#creates empty list, iterate over each patient in data set, statistical analysis for mean and standard deviation
dementiaPatient = []
noDementiaPatient = []
for patient in Patient.filter(Patient.all_patients, diagnosis = "Dementia"):
    dementiaPatient.append(patient.educationyears)
for patient in Patient.filter(Patient.all_patients, diagnosis = "No dementia"):
    noDementiaPatient.append(patient.educationyears)
print (dementiaPatient)
print(noDementiaPatient)
dementia_mean = (statistics.mean(dementiaPatient))
noDementia_mean = (statistics.mean(noDementiaPatient))
dementia_stdev = (statistics.stdev(dementiaPatient))
noDementia_stdev = (statistics.stdev(noDementiaPatient))
Patient_dementia_cols = ['Dementia', 'No Dementia']
mean_age = [dementia_mean, noDementia_mean]
stdev_age = [dementia_stdev, noDementia_stdev]
yerr = [np.zeros(len(mean_age)), stdev_age]
```

The following code finds and returns the t statistic and p value for the bar graph, which can then be analyzed to determine statistical significance. 

```python
#returns p value and t statistic of data
t_stat, p_val = stats.ttest_ind(dementiaPatient, noDementiaPatient)
print(f't_stat = {t_stat}, p_val = {p_val}')
```

The x and y labels of the bar graph as well as the title must also be plotted and this can be done with the following code. 

```python
#plot above information on bar graph
plt.bar(Patient_dementia_cols, mean_age, yerr=yerr, capsize=10, color=["blue", "orange"])
plt.title("Average Years of Education for Dementia Patients vs. Non-Dementia Patients")
plt.ylabel("Years of Education")
plt.show()
```
